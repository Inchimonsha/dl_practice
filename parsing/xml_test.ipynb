```python
import os
from lxml import etree
```

```python
xml_string = """
<root>
    <person id="1">
        <name>Иван</name>
        <age>30</age>
    </person>
</root>
"""
root = etree.fromstring(xml_string)
print(etree.tostring(root, pretty_print=True, encoding="unicode"))
```

#### Задание:

Создайте XML-документ с информацией о трех книгах (название, автор, год издания) и выведите его на экран.

```python
data_dir_path = r"..\resources\raw"
file_path = os.path.join(data_dir_path, "books.xml")

with open(file_path, "rb") as f:
    data_books = f.read()
```

```python
root = etree.fromstring(data_books)
print(etree.tostring(root, pretty_print=True, encoding="unicode"))
```

### Парсинг XML из файла

```python
tree = etree.parse(file_path)
root = tree.getroot()

for book in root.findall("book"):
    title = book.find("title").text
    author = book.find("author").text
    print(f"Книга: {title}, Автор: {author}")
```

### Поиск элементов и работа с атрибутами

```python
data_dir_path = r"..\resources\raw"
file_path = os.path.join(data_dir_path, "persons.xml")

tree = etree.parse(file_path)
root = tree.getroot()

for person in root.findall("person"):
    person_id = person.get("id")
    name = person.find("name").text
    print(f"ID: {person_id}, Имя: {name}")
```

#### Задание:

Добавьте в books.xml атрибут category для каждой книги. Напишите скрипт, который выводит книги определенной категории.

```python
data_dir_path = r"..\resources\raw"
file_path = os.path.join(data_dir_path, "books.xml")

tree = etree.parse(file_path)
root = tree.getroot()

category = 'fantasy'

for book in root.findall("book"):
    cat = book.get("category")
    if category == cat:
        name = book.find("title").text
        author = book.find("author").text
        print(f"{name}, {author}")
```

### Использование XPath

```python
# Найти всех авторов книг с ценой > 500
expensive_books = root.xpath("//book[price>500]/author/text()")
print("Авторы дорогих книг:", expensive_books)
```

### Изменение XML-документа

```python
# Добавление новой книги
new_book = etree.SubElement(root, "book")
etree.SubElement(new_book, "title").text = "Новая книга"
etree.SubElement(new_book, "author").text = "Новый автор"
```

```python
new_book = etree.SubElement(root, "book")
etree.SubElement(new_book, "title").text = "Удаляемая книга"
etree.SubElement(new_book, "author").text = "Новый автор"
```

```python
# Удаление книги
for book in root.findall("book"):
    if book.find("title").text == "Удаляемая книга":
        root.remove(book)
```

```python
# Сохранение изменений
upd_file_path = os.path.join(data_dir_path, "updated_books.xml")
tree.write(upd_file_path, pretty_print=True, encoding="utf-8")
```

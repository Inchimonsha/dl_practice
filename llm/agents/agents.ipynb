Начальный уровень

    Создание простого чат-бота с использованием LLM API
    Задача: написать чат-бота, который отвечает на простые вопросы, используя API OpenAI или аналогичный. Реализовать управление контекстом с помощью ограниченной памяти.
    Цель: понять интеграцию LLM и основы работы с контекстом.

    Автоматизация простого рабочего процесса с n8n
    Задача: настроить рабочий процесс в n8n, который на основе события (например, получение email) инициирует запрос к AI-модели и отправляет ответ на почту.
    Цель: освоить принципы оркестрации и интеграции AI-агента в реальные процессы.

Средний уровень

    Реализация Retrieval-Augmented Generation (RAG)
    Задача: создать AI-агента, который для ответов использует собственный корпус документов, реализовав поиск релевантных данных через векторные базы и дополнение ими генерации.
    Цель: изучить работу с векторными базами, индексацией и расширение контекста.

    Разработка ReAct агента
    Задача: построить агента, который сочетается рассуждение и действия (например, взаимодействие с API, выполнение проверки условий и принятие решений на основе логики).
    Цель: понять алгоритм ReAct и применение логики в цепочках действий.

    Многоагентная система с AutoGen
    Задача: разработать пример с несколькими AI-агентами, которые обмениваются информацией и координируют выполнение задачи (например, анализ текста и составление отчёта).
    Цель: изучить мультиагентное взаимодействие и кооперацию агентов.

Продвинутый уровень

    Интеграция AI-агента с внешними системами через API
    Задача: создать агента, который взаимодействует с несколькими внешними сервисами (CRM, база данных, веб-сервисы), автоматизируя сложный бизнес-процесс.
    Цель: отработать навыки интеграции, управления состоянием и безопасности.

    Разработка автономного AI-агента с возможностью самообучения
    Задача: построить агента, который анализирует свои ошибки, собирает обратную связь и улучшает свои модели или правила поведения без постоянного человеческого участия.
    Цель: освоить методы обучения с подкреплением и активного обучения.

    Создание сложного оркестратора рабочих процессов с n8n и AI
    Задача: разработать систему автоматизации, которая сочетается несколько агентов и внешних API, с динамическим управлением задачами, триггерами и условиями выполнения.
    Цель: понять сложные сценарии управления и масштабирования AI-агентов.

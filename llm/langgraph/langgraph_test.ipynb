```python
from langchain.chains import LLMChain
from langchain.chat_models import ChatOpenAI
from langgraph.prebuilt import create_react_agent
```

```python
from typing import Sequence, Union, Dict, Any, Callable, Optional
from langchain_core.tools import BaseTool
from langchain_core.utils import convert_to_openai_tool
from langchain_core.runnable import Runnable, RunnablePassthrough

class ChatOpenRouter(ChatOpenAI):
    def __init__(self, **kwargs):
        OPENROUTER_API_KEY = "sk-or-ваш_ключ"
        super().__init__(
            model_name=kwargs.get("model_name", "openai/gpt-4o"),
            openai_api_key=OPENROUTER_API_KEY,
            base_url="https://openrouter.ai/api/v1",
            **kwargs,
        )
    
    def bind_tools(
        self,
        tools: Sequence[Union[Dict[str, Any], Callable, BaseTool]],
        *,
        tool_choice: Optional[Union[dict, str, bool]] = None,
        **kwargs: Any,
    ) -> Runnable:
        # Преобразуем инструменты в нужный формат
        formatted_tools = [convert_to_openai_tool(tool) for tool in tools]

        # Формируем параметры для runnable (пример, зависит от реализации)
        runnable_params = dict(tools=formatted_tools)
        if tool_choice is not None:
            runnable_params["tool_choice"] = tool_choice
        
        # Возвращаем runnable, который может быть дальше использован в create_react_agent
        # Здесь можно вернуть self, если он Runnable, или обернуть в RunnablePassthrough
        return RunnablePassthrough(self)
```

```python
# Здесь определяется калькулятор и генератор кода как функции/инструменты (пример через функцию)
def calculator_tool(query: str) -> str:
    """
    Инструмент калькулятора: вычисляет арифметическое выражение,
    переданное в query, безопасно с обработкой ошибок.
    """
    try:
        return str(eval(query))
    except Exception as e:
        return f"Ошибка: {e}"
    
tools = [calculator_tool]

llm = ChatOpenRouter()

agent = create_react_agent(llm, tools)

response = agent.ainvoke({'input': "Вычисли число Фибоначчи 10"})
print(response)
```

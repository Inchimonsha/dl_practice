

    Создание графа состояний с узлами, представляющими разные этапы работы агента (например, ввод сообщения, вызов LLM, обработка результатов).



    Реализация многошагового workflow агента с условными переходами, циклами и обработкой ошибок.




    Разработка мультиагентной системы с параллельным выполнением нескольких веток графа.

    Создание агента с персистентной памятью и отслеживанием состояния с помощью checkpointer.

    Интеграция LangChain LLM-узлов в граф для сложного взаимодействия (например, чат-бот, который может обращаться к разным инструментам в зависимости от контекста).

    Реализация агента для сбора новостей с использованием инструментов (RSS, веб-скрейпер) и генерации итоговой статьи.

    Отладка и мониторинг поведения графа с использованием встроенных инструментов LangGraph.

```python
# Пример простого графа с LangGraph и вызовом ChatOpenAI из LangChain

from langchain_openai import ChatOpenAI
from langgraph.graph import START, StateGraph
from langgraph.checkpoint.memory import MemorySaver
from langchain_core.messages import HumanMessage

# Инициализация LLM
llm = ChatOpenAI(model_name="gpt-4o-mini", temperature=0)

# Создаем граф состояний
workflow = StateGraph(state_schema=dict)

# Функция для узла с вызовом LLM
def call_model(state):
    response = llm.invoke(state["messages"])
    return {"messages": response}

# Определяем граф
workflow.add_edge(START, "model")
workflow.add_node("model", call_model)

# Добавляем память
memory = MemorySaver()
app = workflow.compile(checkpointer=memory)

# Запускаем
config = {"configurable": {"thread_id": "sample-conversation-1"}}
output = app.invoke({"messages": [HumanMessage("Привет! Как дела?")]}, config)
print(output["messages"][-1].pretty_print())
```

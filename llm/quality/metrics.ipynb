# Оценка качества больших языковых моделей (LLM)

### Задание 7: Метрики для оценки LLM

Описание:
Вы оцениваете качество генерации текста большой языковой моделью (например, GPT-3, Llama, Mistral).
Требования:

Выберите задачу (например, генерация ответов на вопросы, суммаризация текста).
Оцените модель с помощью метрик:

BLEU (для оценки точности перевода или генерации)
ROUGE (для оценки суммаризации)
Perplexity (для оценки качества языковой модели)
Human Evaluation (например, оценка людьми по шкале от 1 до 5)

Сравните результаты модели на разных задачах.
Проанализируйте, в каких случаях автоматические метрики (BLEU, ROUGE) могут давать некорректную оценку.
Дополнительно:

Попробуйте использовать метрику BERTScore для более точной оценки качества текста.


### Задание 8: Сравнение LLM на задаче вопрос-ответ

Описание:
Вы сравниваете несколько LLM (например, GPT-3, Llama, Mistral) на задаче ответов на вопросы.
Требования:

Используйте датасет SQuAD.
Оцените модели по метрикам:

Exact Match (EM)
F1-score (для оценки точности ответов)

Проанализируйте, какая модель лучше справляется с:

Фактическими вопросами
Вопросами, требующими рассуждений

Оцените скорость генерации ответов.
Дополнительно:

Проведите A/B-тестирование с участием людей для субъективной оценки качества ответов.

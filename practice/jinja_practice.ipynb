````python
from pydantic import BaseModel, Field
from typing import List
from jinja2 import Template
import json
````

````python
class FillBlock(BaseModel):
    content: str = Field(default="# ВАШ КОД ЗДЕСЬ")

class ExpectedOutput(BaseModel):
    content: str

class Task(BaseModel):
    id: int
    title: str
    description: str
    difficulty: str
    fill_block: FillBlock
    expected_output: ExpectedOutput

class TaskDocument(BaseModel):
    title: str
    topic: str
    level: int
    max_level: int
    tasks: List[Task]
````

````python
TASK_TEMPLATE = Template("""
    # {{ title }}
    **Тема:** {{ topic }} | **Уровень:** {{ level }}/{{ max_level }}

    {% for task in tasks %}
    ## Задание {{ task.id }}: {{ task.title }}

    **Сложность:** {{ task.difficulty.title() }}

    **Условие:**
    {{ task.description }}

    **Ваше решение:**
    ```
    {{ task.fill_block.content }}
    ```
                         
    {% endfor %}
                         
    <i>Отправьте заполненный файл на проверку командой /check</i>
""")
````

````python
test_json = {
"title": "Практика Python asyncio",
"topic": "Python asyncio",
"level": 5,
"max_level": 10,
"tasks": [
{
"id": 1,
"title": "Асинхронный HTTP-клиент",
"description": "Создайте функцию для параллельного сбора содержимого 3 сайтов с помощью aiohttp. Верните список статусов.",
"difficulty": "medium",
"fill_block": {"content": "# ВАШ КОД ЗДЕСЬ"},
"expected_output": {"content": "['200', '200', '200']"},
},
{
"id": 2,
"title": "Пул асинхронных задач",
"description": "Реализуйте пул из 5 воркеров для обработки списка задач.",
"difficulty": "hard",
"fill_block": {"content": "# ВАШ КОД ЗДЕСЬ"},
"expected_output": {"content": "ANSWER"}
}
]
}
````

````python
# data = json.loads(test_json)
task_doc = TaskDocument.model_validate(test_json)
md_content = TASK_TEMPLATE.render(task_doc.model_dump())
````

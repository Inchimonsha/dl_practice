```python
from gensim.models import Word2Vec

# Пример корпуса — список предложений, каждое разбито на слова
sentences = [
    ['i', 'like', 'apple', 'pie', 'for', 'dessert'],
    ['i', 'dont', 'drive', 'fast', 'cars'],
    ['data', 'science', 'is', 'fun'],
    ['chocolate', 'is', 'my', 'favorite'],
    ['my', 'favorite', 'movie', 'is', 'predator']
]

# Обучение модели Word2Vec
# sg=0 — CBOW, sg=1 — Skip-gram
model = Word2Vec(sentences, vector_size=5, window=3, min_count=1, sg=1)

# Получаем вектор для слова
print("Вектор слова 'chocolate':", model.wv['chocolate'])

# Список слов в словаре модели
print("Слова в словаре:", list(model.wv.index_to_key))

# Поиск похожих слов
print("Похожие слова к 'favorite':", model.wv.most_similar('favorite'))
```

```python
from gensim.models import Word2Vec

# Корпус — список предложений, каждое предложение — список слов
sentences = [
    ['king', 'queen', 'man', 'woman', 'child', 'prince', 'princess'],
    ['man', 'woman', 'child', 'family', 'father', 'mother'],
    ['king', 'queen', 'castle', 'throne', 'royal'],
    ['prince', 'princess', 'royal', 'family'],
    ['man', 'king', 'strong', 'brave'],
    ['woman', 'queen', 'wise', 'kind'],
    ['child', 'play', 'fun', 'happy'],
    ['father', 'mother', 'family', 'love']
]

# Обучаем модель Word2Vec (skip-gram, размер вектора 50, окно 3)
model = Word2Vec(sentences, vector_size=50, window=3, min_count=1, sg=1, epochs=100)

# Сохраняем модель (опционально)
model.save("word2vec_example.model")

# Загружаем модель (если нужно)
# model = Word2Vec.load("word2vec_example.model")

# Аналогия: king - man + woman = ?
result = model.wv.most_similar(positive=['king', 'woman'], negative=['man'], topn=1)

print("king - man + woman =", result)
```

```python
import nltk
from nltk.corpus import stopwords
import spacy

# Проверяем и загружаем стоп-слова, если нужно
try:
    nltk.data.find('corpora/stopwords')
except LookupError:
    nltk.download('stopwords')

# Загружаем модель SpaCy для русского языка
nlp = spacy.load("ru_core_news_md")

def preprocess(text):
    stop_words = set(stopwords.words('russian'))
    doc = nlp(text.lower())
    tokens = [token.text for token in doc if token.is_alpha and token.text not in stop_words]
    return tokens

text = "Пример текста на русском языке."
print(preprocess(text))
```

```python
import spacy
nlp = spacy.load("en_core_web_sm")
vector_cat = nlp("cat").vector
vector_dog = nlp("dog").vector
similarity = vector_cat.dot(vector_dog) / ( (vector_cat**2).sum()**0.5 * (vector_dog**2).sum()**0.5 )
print("Similarity between 'cat' and 'dog':", similarity)
```

```python
import spacy
nlp = spacy.load("en_core_web_sm")
text = "Natural language processing is a field of study focused on making sense of text data."
doc = nlp(text)
keywords_freq = [token.text for token in doc if not token.is_stop and token.is_alpha]
keywords_semantic = [token.text for token in doc if not token.is_stop and token.vector_norm > 0]
print("Keywords based on frequency:", keywords_freq)
print("Keywords based on semantics:", keywords_semantic)
```

```python
import spacy

import textacy.extract

nlp = spacy.load('en_core_web_lg')

text = ("London, the capital of England, is not only the most populous city "
        "in the UK but also a leading tourist destination.")

doc = nlp(text)

# 1. Извлечение именованных сущностей
print("Named Entities:")
for ent in doc.ents:
    print(f" - {ent.text} ({ent.label_})")

# 2. Поиск грамматических паттернов
pattern = [{"LOWER": "london"}, {"LEMMA": "be"}, {"DEP": "attr"}]
print("\nMatches:")
for match in textacy.extract.match_patterns(doc, patterns=[pattern]):
    print(f" - {match.text}")

# 3. Извлечение полуструктурированных утверждений
print("\nSemistructured Statements:")
statements = textacy.extract.semistructured_statements(doc, entity="London", cue="is")
for statement in statements:
    subject, verb, fact = statement
    print(f" - {fact}")
```

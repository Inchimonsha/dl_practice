```python
import nltk
nltk.download("stopwords") # Скачиваем список стоп-слов
nltk.download('punkt')      # Скачиваем пакет для токенизации
nltk.download('punkt_tab')

from nltk.corpus import stopwords
from nltk.tokenize import word_tokenize

worf_quote = "Sir, I protest. I am not a merry man!" # Пример текста для обработки

words_in_quote = word_tokenize(worf_quote) # Токенизируем текст по словам
# Результат: ['Sir', ',', 'protest', '.', 'merry', 'man', '!']

stop_words = set(stopwords.words("english")) # Загружаем набор стоп-слов для английского языка

filtered_list = [word for word in words_in_quote if word.casefold() not in stop_words] # Фильтруем слова, исключая стоп-слова

# print(filtered_list)
```

```python
!pip install svgling

import nltk
from nltk.tokenize import word_tokenize
from nltk.tree import Tree
import svgling
from IPython.display import display

# Скачиваем необходимые пакеты
nltk.download('punkt')
nltk.download('averaged_perceptron_tagger')

lotr_quote = "It's a dangerous business, Frodo, going out your door."

# Токенизация
words_in_lotr_quote = word_tokenize(lotr_quote)

# POS-тегирование
lotr_pos_tags = nltk.pos_tag(words_in_lotr_quote)

# Грамматика для именных фраз
grammar = "NP: {<DT>?<JJ>*<NN>}"

# Парсер чанков
chunk_parser = nltk.RegexpParser(grammar)
tree = chunk_parser.parse(lotr_pos_tags)

# Визуализация дерева с помощью svgling
svgling.draw_tree(tree)

# Или просто
# display(tree)
```

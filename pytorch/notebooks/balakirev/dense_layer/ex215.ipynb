```python
import torch
import torch.nn as nn

torch.manual_seed(1)

# здесь объявляйте класс модели (важно именно здесь)
class Model(nn.Module):
    def __init__(self):
        super().__init__()

        self.layer1 = nn.Linear(10, 64, bias=False)
        self.layer2 = nn.Linear(64, 1, bias=True)

        self.bn = nn.BatchNorm1d(64)

    def forward(self, x):
        y = self.layer1(x).relu()
        y = self.bn(y)
        y = self.layer2(y)
        return y

model = Model() # здесь создавайте объект модели

batch_size = 16
x = torch.rand(batch_size, 10)  # этот тензор в программе не менять

# здесь продолжайте программу
model.eval()
predict = model(x)
predict
```

```python
import torch
import torch.utils.data as data
import torch.nn as nn
import torch.optim as optim

from sklearn.datasets import load_breast_cancer

data_breast_cancer = load_breast_cancer()

_global_var_data_x = torch.tensor(data_breast_cancer.data, dtype=torch.float32)
_global_var_target = torch.tensor(data_breast_cancer.target, dtype=torch.int64)

# здесь продолжайте программу
class Model(nn.Module):
    def __init__(self):
        super().__init__()

        self.layer1 = nn.Linear(30, 32, bias=False)
        self.layer2 = nn.Linear(32, 20, bias=False)
        self.layer3 = nn.Linear(20, 1)

        self.bn1 = nn.BatchNorm1d(32)
        self.bn2 = nn.BatchNorm1d(20)

    def forward(self, x):
        y = self.layer1(x).relu()
        y = self.bn1(y)
        y = self.layer2(y).relu()
        y = self.bn2(y)
        y = self.layer3(y)
        return y
    
model = Model()

ds = data.TensorDataset(_global_var_data_x, _global_var_target.float())

d_train, d_test = data.random_split(dataset=ds, lengths=[0.7, 0.3])
train_data = data.DataLoader(d_train, batch_size=16, shuffle=True)
test_data = data.DataLoader(d_test, batch_size=len(d_test), shuffle=False)

lr = 0.01
epochs = 5

optimizer = optim.Adam(model.parameters(), lr=lr, weight_decay=0)
loss_func = nn.BCEWithLogitsLoss()

model.train()

for _e in range(epochs):
    for x_train, y_train in train_data:
        predict = model(x_train)
        loss = loss_func(predict.squeeze(), y_train)

        optimizer.zero_grad()
        loss.backward()
        optimizer.step()

model.eval()

p = 0
for x_val, y_val in test_data:
    predict = model(x_val).squeeze()
    p_class = (torch.sigmoid(predict) > 0.5).long()
    p += (p_class == y_val.long()).sum().item()

Q = p / len(d_test)
Q
```
